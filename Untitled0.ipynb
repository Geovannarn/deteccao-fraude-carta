```python
def avaliar_modelo(nome, y_true, y_pred, y_proba):
    print(f"===== {nome} =====")
    print(classification_report(y_true, y_pred, target_names=['Normal', 'Fraude']))

    cm = confusion_matrix(y_true, y_pred)
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
                xticklabels=['Normal', 'Fraude'], yticklabels=['Normal', 'Fraude'])
    plt.title(f'Matriz de Confusão - {nome}')
    plt.ylabel('Real')
    plt.xlabel('Previsto')
    plt.show()

    fpr, tpr, _ = roc_curve(y_true, y_proba)
    auc = roc_auc_score(y_true, y_proba)
    plt.plot(fpr, tpr, label=f'{nome} (AUC = {auc:.4f})')
    plt.plot([0, 1], [0, 1], linestyle='--', color='gray')
    plt.xlabel('Falso Positivo')
    plt.ylabel('Verdadeiro Positivo')
    plt.title(f'Curva ROC - {nome}')
    plt.legend()
    plt.show()

    prec, rec, _ = precision_recall_curve(y_true, y_proba)
    ap = average_precision_score(y_true, y_proba)
    plt.plot(rec, prec, label=f'{nome} (AP = {ap:.4f})')
    plt.xlabel('Recall')
    plt.ylabel('Precisão')
    plt.title(f'Curva Precisão-Recall - {nome}')
    plt.legend()
    plt.show()

    return {
        'modelo': nome,
        'recall_fraude': recall_score(y_true, y_pred),
        'precisao_fraude': precision_score(y_true, y_pred),
        'f1_fraude': f1_score(y_true, y_pred),
        'auc_roc': auc,
        'auc_pr': ap
    }

resultados = []
```
